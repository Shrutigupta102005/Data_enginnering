```python
import cv2
import os

# Get the current working directory
current_dir = os.getcwd()
print(f"Current working directory: {current_dir}")

# Construct the full path to the image
image_path = os.path.join(current_dir, 'open cv', 'shruti.jpg')
print(f"Full image path: {image_path}")

# Check if the file exists
if not os.path.exists(image_path):
    print(f"Error: The file {image_path} does not exist.")
else:
    # Loading an image
    img = cv2.imread(image_path, 0)
    
    if img is None:
        print(f"Error: Unable to read the image at {image_path}")
    else:
        print(f"Image shape before resize: {img.shape}")
        
        # Resize the image
        img = cv2.resize(img, (400, 400))
        
        print(f"Image shape after resize: {img.shape}")

        # Displaying the image
        cv2.imshow('Image', img)
        cv2.waitKey(0)  # 0 means wait indefinitely
        cv2.destroyAllWindows()
```
